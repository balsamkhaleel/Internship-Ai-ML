# Day 1 — Descriptive Statistics 

Computing and interpreting the center and spread of the Fare column from the Titanic dataset.

## Step 1 — Load and Inspect the Dataset

```python
import pandas as pd

df = pd.read_csv ("tested.csv")

# Information on the dataset
print(df.shape)
df.info()
```

### Missing Values
Check how many values are missing per column.

```python
df.isnull().sum()
```

Dropped the one row with a missing Fare value, since only a single value is missing out of 418 rows — too small to affect the statistics either way.

```python
df = df.dropna(subset=["Fare"])
```

## Step 2 — Compute Descriptive Statistics
Select the Fare column and compute its mean, median, mode, standard deviation, and IQR.

```python
fare_col = df["Fare"]
```

```python
mean_val = fare_col.mean()
print (f"Mean = ", mean_val)

median_val= fare_col.median()
print (f"Median = ",median_val )

mode_val = fare_col.mode()[0]
print(f"Mode = " , mode_val)

std_val =fare_col.std()
print(f"Standard Deviation = ",std_val) 

q1 = fare_col.quantile(0.25)
q3 = fare_col.quantile(0.75)

IQR = q3 - q1
print (f"IQR = ", IQR)
```

## Step 3 — Mean vs. Median: Which Represents a Typical Fare?

The mean (35.63) and median (14.45) are far apart — a difference of about 21. This large gap signals that the Fare column is heavily skewed by a small number of passengers who paid very high fares, which pulls the mean well above where most fares actually fall.

For this reason, the median is the more honest representation of a "typical" fare. The mean is distorted by outliers, while the median reflects the fare that a passenger in the middle of the distribution actually paid.
