# Day 3 — Linear Algebra for ML

Representing data as vectors and matrices, computing predictions with the dot product, and understanding matrix multiplication's shape-matching rule.

## Step 1 — Representing Data as a Matrix
Represent three data samples (age, salary, id_number) as a (3 × 3) NumPy matrix — one row per sample, one column per feature.

```python
import numpy as np
#feature age , salary , id_number
array = np.array([[23,4000,5],
                 [34,6000,2],
                 [44, 7500,8]])
print(array.shape)
```

The shape (3, 3) confirms 3 samples with 3 features each, matching the (rows, columns) = (samples, features) convention.

## Step 2 — Dot Product by Hand and with np.dot
Compute the dot product of one sample's feature vector with a weight vector, then verify the hand-computed result using np.dot.

```python
feature = np.array([34,4000,2])
weights = np.array([0.2,0.04,0.03])
prediction1 = feature @ weights
print(prediction1)

# verify
prediction2 = np.dot(feature ,weights )
print(prediction2)
```

Both computations agree: feature @ weights and np.dot(feature, weights) are two ways of writing the exact same operation — this is exactly how a linear model turns one sample's features into a prediction.

## Step 3 — Matrix Multiplication for All Samples at Once
Use matrix multiplication (@) to produce a prediction for all three samples in a single operation, instead of looping over rows.

```python
prediction3= array @ weights
print(prediction3)
```

The result is a vector of 3 predictions — one per sample — produced in a single @ operation. This is why matrix multiplication matters: it applies the dot product across an entire dataset at once.

## Step 4 — Deliberately Causing a Shape Mismatch
Create a weight vector with the wrong length and try to multiply it with feature to trigger a shape-mismatch error on purpose.

```python
weight = np.array([0.4,0.5])

prediction4 = feature @ weight
print (prediction4)
```

**Why this error happened:**

feature has shape (3,) — it holds 3 numbers (age, salary, id_number). weight was created with shape (2,) — only 2 numbers. Matrix multiplication requires the inner dimensions to match: for A @ B, the number of columns in A must equal the number of elements in B. Here, feature has 3 "columns" but weight only has 2, so NumPy raises ValueError: ... size 2 is different from 3.



## Summary
Represented a dataset as a matrix, computed a prediction with the dot product (verified two equivalent ways), used matrix multiplication to predict for all samples at once, and deliberately triggered and explained a shape-mismatch error — the most common bug when working with vectors and matrices in ML.
