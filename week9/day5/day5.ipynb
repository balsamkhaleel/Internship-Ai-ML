## 2. Technical Write-Up

### Problem Statement

Cardiovascular disease is one of the major health challenges worldwide. Early identification of cardiovascular risk can support timely medical evaluation and preventive care.

This project develops a **Cardiac Patient Monitoring System** that uses machine learning to estimate cardiovascular risk based on patient health information such as age, blood pressure, cholesterol, glucose level, lifestyle factors, and other clinical attributes.

The system is designed as an educational decision-support project and is **not intended to provide a medical diagnosis or replace professional medical evaluation**.

---

### Dataset

The project uses the **Cardiovascular Disease Dataset**, which originally contains approximately **70,000 patient records**.

The dataset includes demographic, clinical, and lifestyle features such as:

- Age
- Gender
- Height
- Weight
- Systolic blood pressure
- Diastolic blood pressure
- Cholesterol
- Glucose
- Smoking status
- Alcohol intake
- Physical activity

The target variable `cardio` indicates the presence or absence of cardiovascular disease.

After data cleaning and preprocessing, invalid and unrealistic observations were removed and additional health-related features were prepared for modeling.

---

### Methodology

The complete machine learning workflow included:

1. Data cleaning and preprocessing
2. Exploratory Data Analysis (EDA)
3. Feature engineering
4. Baseline and supervised machine learning models
5. Deep learning using a Neural Network
6. Hyperparameter tuning
7. Classification threshold optimization
8. Model evaluation and error analysis
9. Model explainability using SHAP
10. Model serialization
11. Experiment tracking using MLflow
12. Streamlit application development
13. Public deployment using Render

The final deployed model is a **Neural Network** with the following tracked configuration:

- Architecture: `64-32-1`
- Learning rate: `0.0001`
- Dropout: `0.0`
- Batch size: `64`
- Classification threshold: `0.42`
- Random seed: `42`

---

### Final Model Results

The final model evaluation recorded in MLflow produced:

| Metric | Result |
|---|---:|
| Accuracy | 0.7187 |
| Precision | 0.6931 |
| Recall | 0.7744 |
| F1-score | **0.7315** |

The **F1-score** was treated as an important evaluation metric because the project requires a balance between identifying cardiovascular-risk cases and controlling incorrect predictions.

Recall was also monitored because false negatives are particularly important in a risk-screening context.

---

### Explainability

SHAP was used to analyze how individual features influenced the model predictions.

Important features investigated during the explainability analysis included:

- `ap_hi`
- `age_years`
- `pulse_pressure`
- `map`

This analysis improved the interpretability of the model and helped explain how important patient characteristics contributed to its predictions.

---

### Deployment

The final model and preprocessing scaler were serialized and integrated into a Streamlit application.

The application was deployed publicly using **Render**:

https://cardiac-patient-monitoring.onrender.com/

Local and deployed predictions were compared using the same patient inputs to verify that the production application reproduced the expected model behavior.

---

### Limitations & Future Improvements

The current system has several limitations:

- The model was trained on a public dataset and may not represent every patient population.
- Predictions depend on the quality and accuracy of the entered patient information.
- The model provides a risk estimate rather than a clinical diagnosis.
- Additional external validation would be required before considering any real clinical use.

Future improvements could include:

- Validation using additional datasets
- Further model optimization
- Improved probability calibration
- Additional explainability features in the user interface
- More extensive deployment monitoring and automated testing

## 5. Full-Project Retrospective

The Cardiac Patient Monitoring System was developed incrementally across four sprints. Each sprint contributed a different stage of the machine learning lifecycle and helped transform the initial dataset into a complete deployed application.

---

### Sprint 1 — Data Understanding & Initial Modeling

Sprint 1 focused on understanding the problem and establishing the foundation of the project.

Key work included:

- Loading and understanding the cardiovascular dataset.
- Cleaning invalid and unrealistic observations.
- Performing Exploratory Data Analysis (EDA).
- Investigating feature distributions and relationships.
- Preparing the data for machine learning.
- Building initial baseline models.
- Establishing the first benchmark for later comparison.

**Main lesson:**  
A reliable machine learning system starts with understanding and preparing the data correctly. Model performance cannot compensate for poor-quality input data.

---

### Sprint 2 — Model Development & Optimization

Sprint 2 focused on improving the predictive modeling process and comparing different approaches.

Key work included:

- Training multiple supervised machine learning models.
- Exploring deep learning with Neural Networks.
- Performing hyperparameter tuning.
- Comparing different model configurations.
- Optimizing the classification threshold.
- Evaluating models using Precision, Recall, and F1-score.
- Selecting a model configuration suitable for further development.

**Main lesson:**  
Model selection should not depend on Accuracy alone. The evaluation metric must reflect the actual objective of the problem, which made F1-score and Recall particularly important during this project.

---

### Sprint 3 — Evaluation, Error Analysis & Explainability

Sprint 3 focused on understanding the behavior and reliability of the selected model.

Key work included:

- Performing detailed model evaluation.
- Analyzing false positives and false negatives.
- Investigating classification errors.
- Applying SHAP for model explainability.
- Examining the influence of important features such as:
  - `ap_hi`
  - `age_years`
  - `pulse_pressure`
  - `map`
- Preparing the final model pipeline for deployment.

**Main lesson:**  
A model should not only generate predictions. Its errors and prediction behavior should also be investigated and documented, especially when working with health-related data.

### Sprint 4 — MLOps, Application & Deployment

Sprint 4 focused on transforming the final model into a reproducible and accessible application.

Key work included:
